```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

pd.set_option("display.max_columns", 40)
```

load dataset

```python
df = pd.read_csv("/content/drive/MyDrive/Data Saya/semester 5/pembelajaran mesin/wbc.csv")
print("Ukuran data:", df.shape)
df.head()
```

1. pisahkan variabel yang dapat digunakan dan tidak dapat digunakan

Dari hasil df.info() di atas, dapat diidentifikasi:

| Kolom | Keterangan | Status |
|---|---|---|
| id | Nomor identitas pasien, tidak punya nilai prediktif | Tidak dapat digunakan atau dibuang |
| Unnamed: 32 | Kolom kosong (seluruh nilainya NaN), muncul akibat trailing comma saat data di-export ke CSV | Tidak dapat digunakan atau dibuang |
| diagnosis | Label kelas (M/B) | Variabel target |
| 30 kolom lainnya (radius_mean ... fractal_dimension_worst) | Hasil pengukuran numerik dari citra sel | Variabel fitur |

```python
# Cek kolom yang seluruhnya kosong
print("Jumlah nilai kosong per kolom (khusus yang > 0):")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
#membuang kolom yang tidak dapat digunakan
kolom_tidak_terpakai = ["id", "Unnamed: 32"]
kolom_tidak_terpakai = [c for c in kolom_tidak_terpakai if c in df.columns]
print("Kolom yang dibuang:", kolom_tidak_terpakai)

df = df.drop(columns=kolom_tidak_terpakai)

target_col = "diagnosis"
fitur_cols = [c for c in df.columns if c != target_col]

print("Kolom target :", target_col)
print("Jumlah kolom fitur:", len(fitur_cols))
print("Daftar kolom fitur:", fitur_cols)
```

```python
X = df[fitur_cols].copy()
y = df[target_col].copy()

print("Ukuran X (fitur):", X.shape)
print("Ukuran y (target):", y.shape)
```

2. Encoding kolom diagnosis

```python
print("Distribusi kelas sebelum encoding:")
print(y.value_counts())
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(y)

print("Mapping label hasil encoding:")
for kelas, kode in zip(le.classes_, le.transform(le.classes_)):
    print(f"  {kelas} -> {kode}")

y_encoded[:10]
```

3. Standarisasi kolom numerik

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=fitur_cols, index=X.index)

X_scaled.describe().loc[["mean", "std"]].round(3)
```

```python
X_scaled.head()
```

4. Stratified Split Data Latih (80%) dan Data Uji (20%)

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y_encoded,
    test_size=0.2,
    stratify=y_encoded,
    random_state=42,
)

print("Ukuran data latih :", X_train.shape)
print("Ukuran data uji   :", X_test.shape)
```

```python
print("Proporsi kelas pada data asli:")
print(pd.Series(y_encoded).value_counts(normalize=True).round(4))

print("\nProporsi kelas pada data latih:")
print(pd.Series(y_train).value_counts(normalize=True).round(4))

print("\nProporsi kelas pada data uji:")
print(pd.Series(y_test).value_counts(normalize=True).round(4))
```
