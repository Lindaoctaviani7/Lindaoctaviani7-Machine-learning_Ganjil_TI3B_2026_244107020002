Import library

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

1. Data Loading and Initial Cleaning

```python
# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/Data Saya/semester 5/pembelajaran mesin/wbc.csv')

# Drop the 'id' column and 'Unnamed: 32' if they exist
if 'id' in df.columns:
    df = df.drop('id', axis=1)
if 'Unnamed: 32' in df.columns:
    df = df.drop('Unnamed: 32', axis=1)

print("Dataset setelah penghapusan kolom awal:")
print(df.head())
```

2. Encoding the 'diagnosis' Column

```python
# Encode the 'diagnosis' column
label_encoder = LabelEncoder()
df['diagnosis'] = label_encoder.fit_transform(df['diagnosis'])

print("\nDataset setelah encoding diagnosis:")
print(df.head())
print("\nDistribusi nilai diagnosis setelah encoding:")
print(df['diagnosis'].value_counts())
```

3. Standardisasi Kolom Numerik

```python
# Identify numerical columns for standardization (exclude 'diagnosis')
numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
if 'diagnosis' in numerical_cols:
    numerical_cols.remove('diagnosis')

# Apply StandardScaler to numerical columns
scaler = StandardScaler()
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

print("\nDataset setelah standardisasi:")
print(df.head())
```
