import library

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
import tensorflow as tf
```

load dataset

```python
# Load dataset
iris = load_iris()
X = iris.data
y = iris.target.reshape(-1, 1)

# One-hot encoding
encoder = OneHotEncoder(sparse_output=False)
y = encoder.fit_transform(y)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Bangun model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

kompilasi

```python
# Kompilasi
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

latih model

```python
# Latih model
model.fit(X_train, y_train, epochs=50, batch_size=8)
```

evaluasi

```python
# Evaluasi
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi: {acc}")
```

Tugas 2

Ubah jumlah neuron hidden layer.

Bandingkan akurasi dengan konfigurasi awal.

```python
# Bangun model baru dengan jumlah neuron hidden layer yang diubah
# Konfigurasi awal: 10 -> 8 -> 3
# Konfigurasi baru: 16 -> 12 -> 3
model_baru = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(12, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])

# Kompilasi model baru
model_baru.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model baru
model_baru.fit(X_train, y_train, epochs=50, batch_size=8, verbose=0)

# Evaluasi model baru
loss_baru, acc_baru = model_baru.evaluate(X_test, y_test, verbose=0)

# membandingkan hasil akurasi
print("Perbandingan Hasil Akurasi")
print(f"Akurasi Awal (10 & 8 neuron) : {acc:.4f}")
print(f"Akurasi Baru (16 & 12 neuron): {acc_baru:.4f}")
```

Tugas 3

Bandingkan Sigmoid vs ReLU pada dataset Iris.

Catat perbedaan loss dan akurasi.

```python
# Model dengan aktivasi ReLU
model_relu = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    tf.keras.layers.Dense(10, activation='relu'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
model_relu.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Model dengan aktivasi Sigmoid
model_sigmoid = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    tf.keras.layers.Dense(10, activation='sigmoid'),
    tf.keras.layers.Dense(8, activation='sigmoid'),
    tf.keras.layers.Dense(3, activation='softmax')
])
model_sigmoid.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih kedua model
# Melatih model dengan ReLU
model_relu.fit(X_train, y_train, epochs=50, batch_size=8, verbose=0)

# Melatih model dengan Sigmoid
model_sigmoid.fit(X_train, y_train, epochs=50, batch_size=8, verbose=0)

# Evaluasi kedua model
loss_relu, acc_relu = model_relu.evaluate(X_test, y_test, verbose=0)
loss_sigmoid, acc_sigmoid = model_sigmoid.evaluate(X_test, y_test, verbose=0)

# menampilkan perbandingan
print("Perbandingan Hasil: ReLU vs Sigmoid")
print(f"Model ReLU    -> Loss: {loss_relu:.4f}, Akurasi: {acc_relu:.4f}")
print(f"Model Sigmoid -> Loss: {loss_sigmoid:.4f}, Akurasi: {acc_sigmoid:.4f}")
```
