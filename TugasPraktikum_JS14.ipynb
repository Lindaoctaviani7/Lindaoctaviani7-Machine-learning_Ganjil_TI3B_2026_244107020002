nomor 1: Modifikasi Model CNN CIFAR-10 untuk Akurasi >80%

Keterangan: Menggunakan Batch Normalization, Dropout, dan penambahan

filter konvolusi untuk meningkatkan kapasitas model tanpa overfitting.

import library

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
```

Memuat dan menormalisasi ulang dataset CIFAR-10

```python
(cifar_train_img, cifar_train_lbl), (cifar_test_img, cifar_test_lbl) = datasets.cifar10.load_data()
cifar_train_img, cifar_test_img = cifar_train_img / 255.0, cifar_test_img / 255.0
```

Membangun arsitektur CNN yang ditingkatkan

```python
model_cifar = models.Sequential([
    # Blok Konvolusi 1
    layers.Conv2D(32, (3, 3), padding='same', activation='relu', input_shape=(32, 32, 3)),
    layers.BatchNormalization(),
    layers.Conv2D(32, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    # Blok Konvolusi 2
    layers.Conv2D(64, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.Conv2D(64, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.3),

    # Blok Konvolusi 3
    layers.Conv2D(128, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.4),

    # Classifier (Fully Connected Layer)
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(10)
])
```

Compile model dengan optimizer Adam

```python
model_cifar.compile(optimizer='adam',
                    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
                    metrics=['accuracy'])

print("Melatih Model CIFAR-10 (Modifikasi)")
history_cifar = model_cifar.fit(cifar_train_img, cifar_train_lbl, epochs=15,
                                validation_data=(cifar_test_img, cifar_test_lbl))
```

Evaluasi performa akhir model CIFAR-10

```python
test_loss_c, test_acc_c = model_cifar.evaluate(cifar_test_img, cifar_test_lbl, verbose=2)
print(f"Akurasi Akhir CIFAR-10 setelah modifikasi: {test_acc_c * 100:.2f}%\n")
```

nomor 2: Membuat Model CNN untuk Klasifikasi Dataset MNIST


Dataset grayscale 28x28 piksel dengan target 10 kelas digit

Memuat dan memproses dataset MNIST

```python
(mnist_train_img, mnist_train_lbl), (mnist_test_img, mnist_test_lbl) = datasets.mnist.load_data()
```

Reshape agar memiliki dimensi channel (28, 28, 1) dan normalisasi

```python
mnist_train_img = mnist_train_img.reshape((60000, 28, 28, 1)) / 255.0
mnist_test_img = mnist_test_img.reshape((10000, 28, 28, 1)) / 255.0
```

Membangun model CNN untuk MNIST

```python
model_mnist = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10)
])
```

Compile model MNIST

```python
model_mnist.compile(optimizer='adam',
                    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
                    metrics=['accuracy'])

print("Melatih Model CNN untuk MNIST")
history_mnist = model_mnist.fit(mnist_train_img, mnist_train_lbl, epochs=5,
                                validation_data=(mnist_test_img, mnist_test_lbl))
```

Evaluasi model MNIST

```python
# Evaluasi model MNIST
test_loss_m, test_acc_m = model_mnist.evaluate(mnist_test_img, mnist_test_lbl, verbose=2)
print(f"Akurasi Akhir MNIST: {test_acc_m * 100:.2f}%")
```
