import library

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

pra pengolahan data

pra pengolahan data training

```python
	train_datagen = ImageDataGenerator(rescale = 1./255,
	                                   shear_range = 0.2,
	                                   zoom_range = 0.2,
	                                   horizontal_flip = True)
	training_set = train_datagen.flow_from_directory('/content/drive/MyDrive/Data Saya/semester 5/pembelajaran mesin/dataset/training_set',
	                                                 target_size = (64, 64),
	                                                 batch_size = 32,
	                                                 class_mode = 'binary')
```

pra pengolahan data testing

```python
	test_datagen = ImageDataGenerator(rescale = 1./255)
	test_set = test_datagen.flow_from_directory('/content/drive/MyDrive/Data Saya/semester 5/pembelajaran mesin/dataset/test_set',
	                                            target_size = (64, 64),
	                                            batch_size = 32,
	                                            class_mode = 'binary')
```

pembuatan model CNN

inisiasi model CNN

```python
cnn = tf.keras.models.Sequential()
```

pembuatan layer konvolusi 1

```python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu',
                               input_shape=[64, 64, 3]))
```

pembuatan layer konvolusi 2 dan pooling 2

```python
	cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
	cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

flattening

```python
cnn.add(tf.keras.layers.Flatten())
```

fully connected layer 1(input)

```python
cnn.add(tf.keras.layers.Dense(units=128, activation='relu'))
```

fully connected layer 2(output)

```python
cnn.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

compile model CNN

```python
cnn.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

fit CNN

```python
cnn.fit(x = training_set, validation_data=test_set, epochs=25)
```
