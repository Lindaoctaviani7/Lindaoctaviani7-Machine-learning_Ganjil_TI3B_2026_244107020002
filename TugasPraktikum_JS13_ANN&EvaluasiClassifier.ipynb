```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Bangun model JST
model = Sequential([
    # Ubah gambar 28x28 menjadi vektor
    Flatten(input_shape=(28, 28)),
    # Hidden layer 1
    Dense(128, activation='relu'),
    # Hidden layer 2
    Dense(64, activation='relu'),
    # Output layer (10 kelas)
    Dense(10, activation='softmax')
])

# 3. Kompilasi model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# 4. Latih model
model.fit(X_train, y_train, epochs=5, batch_size=32, validation_split=0.1)

# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi pada data uji: {acc:.4f}")
```

```python
import time
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Input

# Fungsi untuk membuat, melatih, dan mengevaluasi model
def jalankan_eksperimen(nama_eksperimen, aktivasi, neuron_list):
    print(f"Memulai Eksperimen: {nama_eksperimen}")

    # 1. Bangun model dengan Input layer sesuai rekomendasi terbaru Keras
    model = Sequential([
        Input(shape=(28, 28)),
        Flatten()
    ])

    # Menambahkan hidden layers sesuai list neuron
    for unit in neuron_list:
        model.add(Dense(unit, activation=aktivasi))

    # Output layer
    model.add(Dense(10, activation='softmax'))

    # 2. Kompilasi model
    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    # 3. Latih model & hitung waktu pelatihan
    mulai_waktu = time.time()
    history = model.fit(
        X_train, y_train,
        epochs=5,
        batch_size=32,
        validation_split=0.1,
        verbose=1
    )
    total_waktu = time.time() - mulai_waktu

    # 4. Evaluasi model
    loss, akurasi = model.evaluate(X_test, y_test, verbose=0)

    print(f"Selesai dalam: {total_waktu:.2f} detik")
    print(f"Akurasi Uji : {akurasi:.4f}\n")

    return total_waktu, akurasi

# menjalankan Eksperimen 1: Model dengan ReLU (3 Hidden Layers: 256, 128, 64)
waktu_relu, akurasi_relu = jalankan_eksperimen(
    nama_eksperimen="ReLU (3 Hidden Layers: 256 -> 128 -> 64)",
    aktivasi="relu",
    neuron_list=[256, 128, 64]
)

# Jalankan Eksperimen 2: Model dengan Sigmoid (3 Hidden Layers: 256, 128, 64)
waktu_sigmoid, akurasi_sigmoid = jalankan_eksperimen(
    nama_eksperimen="Sigmoid (3 Hidden Layers: 256 -> 128 -> 64)",
    aktivasi="sigmoid",
    neuron_list=[256, 128, 64]
)

# 5. Bandingkan Hasil
print("Perbandingan Hasil")
print(f"{'Aktivasi':<15} | {'Waktu Pelatihan (detik)':<25} | {'Akurasi Uji':<12}")
print("-" * 60)
print(f"{'ReLU':<15} | {waktu_relu:<25.2f} | {akurasi_relu:<12.4f}")
print(f"{'Sigmoid':<15} | {waktu_sigmoid:<25.2f} | {akurasi_sigmoid:<12.4f}")
```
