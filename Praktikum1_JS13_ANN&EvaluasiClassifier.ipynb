Langkah:

Buat dataset sederhana (XOR).

Inisialisasi bobot dan bias.

Implementasikan forward pass.

Hitung error dan lakukan backpropagation.

Update bobot menggunakan gradient descent.

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

# Output akhir
print("Prediksi:")
print(a2)
```

Tugas 1:

Ubah jumlah neuron hidden layer menjadi 3.

Bandingkan hasil loss dengan konfigurasi awal.

Tambahkan fungsi aktivasi ReLU dan bandingkan hasil.

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter umum
input_size = 2
output_size = 1
lr = 0.1
epochs = 10000

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1.0, 0.0)

# Fungsi training helper untuk mempermudah perbandingan
def train_nn(hidden_size, activation='sigmoid'):
    # Inisialisasi bobot (menggunakan seed agar konsisten)
    np.random.seed(42)
    W1 = np.random.randn(input_size, hidden_size)
    b1 = np.zeros((1, hidden_size))
    W2 = np.random.randn(hidden_size, output_size)
    b2 = np.zeros((1, output_size))

    loss_history = []

    for epoch in range(epochs):
        # Forward pass
        z1 = np.dot(X, W1) + b1
        if activation == 'sigmoid':
            a1 = sigmoid(z1)
        else: # relu
            a1 = relu(z1)

        z2 = np.dot(a1, W2) + b2
        a2 = sigmoid(z2) # Output layer tetap menggunakan sigmoid karena output biner [0,1]

        # Hitung error
        error = y - a2
        loss = np.mean(np.square(error))
        loss_history.append(loss)

        # Backpropagation
        d_a2 = error * sigmoid_derivative(a2)
        d_W2 = np.dot(a1.T, d_a2)
        d_b2 = np.sum(d_a2, axis=0, keepdims=True)

        if activation == 'sigmoid':
            d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
        else:
            d_a1 = np.dot(d_a2, W2.T) * relu_derivative(a1)

        d_W1 = np.dot(X.T, d_a1)
        d_b1 = np.sum(d_a1, axis=0, keepdims=True)

        # Update bobot
        W1 += lr * d_W1
        b1 += lr * d_b1
        W2 += lr * d_W2
        b2 += lr * d_b2

    return loss_history[-1], a2

# 1. Konfigurasi Awal (Sigmoid, Hidden=2)
loss_init, pred_init = train_nn(hidden_size=2, activation='sigmoid')

# 2. Konfigurasi Baru (Sigmoid, Hidden=3)
loss_h3, pred_h3 = train_nn(hidden_size=3, activation='sigmoid')

# 3. Konfigurasi Baru dengan ReLU (ReLU, Hidden=3)
loss_relu, pred_relu = train_nn(hidden_size=3, activation='relu')

# Bandingkan Hasil
print(" PERBANDINGAN HASIL LOSS AKHIR ")
print(f"1. Konfigurasi Awal (Sigmoid, Hidden=2)      : Loss = {loss_init:.6f}")
print(f"2. Konfigurasi Baru (Sigmoid, Hidden=3)      : Loss = {loss_h3:.6f}")
print(f"3. Konfigurasi Baru dengan ReLU (Hidden=3)   : Loss = {loss_relu:.6f}")
print(" PREDIKSI MASING-MASING MODEL ")
print("Target asli:")
print(y)
print("\nPrediksi 1 (Sigmoid, Hidden=2):")
print(np.round(pred_init, 4))
print("\nPrediksi 2 (Sigmoid, Hidden=3):")
print(np.round(pred_h3, 4))
print("\nPrediksi 3 (ReLU, Hidden=3):")
print(np.round(pred_relu, 4))
```
